# Проект для «Викишоп»

**Описание проекта**

Интернет-магазин «Викишоп» запускает новый сервис. Теперь пользователи могут редактировать и дополнять описания товаров, как в вики-сообществах. То есть, клиенты предлагают свои правки и комментируют изменения других. Магазину нужен инструмент, который будет искать токсичные комментарии и отправлять их на модерацию. 


**Цель**

Реализовать модель для классификации комментариев к описанию товаров на позитивные и негативные. 


**План реализации**

1. Загрузка и анализ данных.
2. Подготовка данных.
3. Обучение моделей и поиск лучшей. 
4. Проверка качества классификации комментариев на тестовой выборке.
5. Формулирование выводов.


**Результат и критерий успеха**

- В результате у нас должна получиться обученная модель, способная классифицировать комментарии к описанию товаров интернет-магазина «Викишоп» на позитивные и негативные
- Значение метрики качества `F1` на тестовой выборке должно быть не меньше `0.75`

Для выполнения проекта применять *BERT* необязательно, но мы можем попробовать.

```python
# создадим функцию для построения диаграмм для категориальных признаков
def barh_and_pie(df, feature, agg_column, agg_func='count'):
    """
    Функция строит столбчатую и круговую диаграммы для абсолютных и относительных значений категориальных признаков 
    (распределение)

    Args:
        df (pd.DataFrame): датафрейм
        feature: категориальный признак
        agg_column: признак для агрегации 
        aggfunc: функция для агрегации (по умолчанию 'count')

    Returns:
        None    
    """
    plt.figure(figsize=(20, 10))
    
    plt.subplot(1, 2, 1)
    plt.barh(df.pivot_table(index=feature, values=agg_column, aggfunc=agg_func).index.astype(str), 
             df.pivot_table(index=feature, values=agg_column, aggfunc=agg_func)[agg_column])
    plt.xlabel(f'Количество')
    plt.title(f'Распределение для признака - "{feature}"', fontsize=12)
    plt.grid(linestyle='--', alpha=0.5)
    

    plt.subplot(1, 2, 2)
    plt.pie(df.pivot_table(index=feature, values=agg_column, aggfunc=agg_func)[agg_column],
            labels=df.pivot_table(index=feature, values=agg_column, aggfunc=agg_func).index, 
            autopct='%1.1f%%') 

    plt.show()
```

### Предобработка текста

```python
# Загрузка предобученной модели английского языка
nlp = spacy.load('en_core_web_sm', disable=['parser', 'ner'])
```

nlp - это пайплайн обработки текста, который включает в себя:
- Токенизатор (разбивает текст на отдельные токены: слова, знаки препинания и т.п.)
- Лемматизатор (приводит слова к базовой форме)
- Определяет часть речи (существительное, глагол, прилагательное и т.п.)
- Определяет стоп-слова (слова без значимой смысловой нагрузки)

```python
# Создадим функцию для предобработки текста с использованием nlp
def preprocess_text(text):
    """
    Функция выполняет предобработку текста с использованием spaCy: 
    - удаляет все символы, кроме букв и пробелов, 
    - приводит к нижнему регистру,
    - разбивает текст на отдельные токены,
    - приводит слова к базовой форме (лемматизирует),
    - удаляет стоп-слова и сохраняет только буквенные токены
    """    
    # очистка текста
    text = re.sub(r'[^a-zA-Z\s]', '', text) # удалим все символы, которые не являются буквами и не являются пробелами
    text = text.lower().strip() # приводим к нижнему регистру
    
    # лемматизация
    lemmatized_text = nlp(text)
    # приведение слов к нормальной форме, удаление стоп-слов и сохранение только буквенных токенов
    tokens = [token.lemma_ for token in lemmatized_text if not token.is_stop and token.is_alpha]
    
    return ' '.join(tokens)
    
```

```python
# Применение предобработки текста
df_toxic_comments['text_preprocessed'] = df_toxic_comments['text'].progress_apply(preprocess_text)
# progress_apply и tqdm автоматически добавляют прогресс-бар
```

### Разделение на обучающую и тестовую выборки

```python
X = df_toxic_comments[['text_preprocessed']]
y = df_toxic_comments['toxic']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, 
                                                    y, 
                                                    test_size=TEST_SIZE, 
                                                    random_state=RANDOM_STATE, 
                                                    stratify=y)
```

### Подготовка Pipeline

Подготовим Pipeline, включающий векторизацию текста с помощью `TfidfVectorizer` и модель

```python
# Создание пайплайна для векторизации текста
preprocessor = ColumnTransformer(
    transformers =[
        ('tfidf', TfidfVectorizer(), 'text_preprocessed')
    ], 
    remainder='passthrough'
)
```

```python
# Общий пайплайн 'pipeline': векторизация текста и модель
def pipeline(model):
    return Pipeline(
        [
            ('preprocessor', preprocessor), 
            ('model', model)
        ]
    )
```

*Большая величина TF-IDF говорит об уникальности слова в тексте по отношению к корпусу. Чем чаще оно встречается в конкретном тексте и реже в остальных, тем выше значение TF-IDF*

### Обучение модели логистической регрессии `LogisticRegression`

```python
lr_pipeline = pipeline(LogisticRegression(class_weight=class_weights_dict, 
                                            max_iter=300, 
                                            random_state=RANDOM_STATE))

# создание словаря гиперпараметров модели
lr_params = {
    #'preprocessor__tfidf__max_features': [20000, None],
    'preprocessor__tfidf__ngram_range': [(1, 1), (1, 2)],
    'model__C': list(np.logspace(-1, 1, 3)), 
    'model__penalty': ['l1', 'l2'], 
    'model__solver': ['liblinear', 'saga']
}
```

```python
%%time
# Перебор гиперпараметров с помощью RandomizedSearchCV
lr_search = (
    RandomizedSearchCV(lr_pipeline, 
                       lr_params, 
                       n_iter=10, 
                       scoring='f1', 
                       cv=CV,
                       random_state=RANDOM_STATE,
                       verbose=2, 
                       n_jobs=-1)
)

lr_search.fit(X_train, y_train)
```

```python
# сохранение результатов RandomizedSearch
results_lr_search = (pd.DataFrame(lr_search.cv_results_)
                     .sort_values(by='rank_test_score')
                     .reset_index(drop=True))

# Сохранение лучшего pipeline из результатов RandomizedSearch
best_lr = lr_search.best_estimator_

# сохранение метрики F1 на кросс-валидации
lr_f1 = lr_search.best_score_
```
